```python
import json
import os

# === ADEGA DE VINHOS COM SALVAMENTO ===
ARQUIVO = "vinhos.json"

# (salvar / carregar)
def carregar_vinhos():
    if os.path.exists(ARQUIVO):
        with open(ARQUIVO, "r", encoding="utf-8") as f:
            return json.load(f)
    else:
        # Dados iniciais
        vinhos_iniciais = [
            {"id": 1, "nome": "WINEhouse Sauvignon", "tipo": "Branco", "ano": 2021, "pais": "Chile"},
            {"id": 2, "nome": "My Chemical Wine Malbec", "tipo": "Tinto", "ano": 2020, "pais": "Argentina"},
            {"id": 3, "nome": "Cemitery Pinot Noir", "tipo": "Tinto", "ano": 2022, "pais": "França"},
            {"id": 4, "nome": "Vampire Blood Merlot", "tipo": "Tinto", "ano": 2021, "pais": "Itália"},
            {"id": 5, "nome": "Ousbourne Tannat", "tipo": "Tinto", "ano": 2019, "pais": "Uruguai"},
        ]
        salvar_vinhos(vinhos_iniciais)
        return vinhos_iniciais

def salvar_vinhos(lista_vinhos):
    with open(ARQUIVO, "w", encoding="utf-8") as f:
        json.dump(lista_vinhos, f, ensure_ascii=False, indent=4)

# --- Carregar dados ---
vinhos = carregar_vinhos()

# --- Funções CRUD ---
def listar_vinhos():
    if not vinhos:
        print("\nNenhum vinho cadastrado ainda.\n")
    else:
        print("\n=== Lista de Vinhos ===")
        for vinho in vinhos:
            print(f"ID: {vinho['id']} | Nome: {vinho['nome']} | Tipo: {vinho['tipo']} | Ano: {vinho['ano']} | País: {vinho['pais']}")
        print()

def cadastrar_vinho():
    print("\n=== Cadastro de Novo Vinho ===")
    nome = input("Nome do vinho: ")
    tipo = input("Tipo (Tinto, Branco, Rosé...): ")
    ano = int(input("Ano da safra: "))
    pais = input("País de origem: ")

    novo_id = max([v["id"] for v in vinhos]) + 1 if vinhos else 1
    vinhos.append({"id": novo_id, "nome": nome, "tipo": tipo, "ano": ano, "pais": pais})
    salvar_vinhos(vinhos)
    print(f"\n✅ Vinho '{nome}' cadastrado com sucesso!\n")

def pesquisar_vinho():
    termo = input("\nDigite o nome (ou parte do nome) do vinho: ").lower()
    resultados = [v for v in vinhos if termo in v["nome"].lower()]
    if resultados:
        print("\n=== Resultados da Pesquisa ===")
        for v in resultados:
            print(f"ID: {v['id']} | Nome: {v['nome']} | Tipo: {v['tipo']} | Ano: {v['ano']} | País: {v['pais']}")
    else:
        print("\n⚠️ Nenhum vinho encontrado com esse nome.\n")

def atualizar_vinho():
    try:
        id_vinho = int(input("\nDigite o ID do vinho que deseja atualizar: "))
        vinho = next((v for v in vinhos if v["id"] == id_vinho), None)
        if vinho:
            print(f"\nAtualizando '{vinho['nome']}'")
            vinho["nome"] = input(f"Novo nome ({vinho['nome']}): ") or vinho["nome"]
            vinho["tipo"] = input(f"Novo tipo ({vinho['tipo']}): ") or vinho["tipo"]
            ano = input(f"Novo ano ({vinho['ano']}): ")
            vinho["ano"] = int(ano) if ano else vinho["ano"]
            vinho["pais"] = input(f"Novo país ({vinho['pais']}): ") or vinho["pais"]
            salvar_vinhos(vinhos)
            print("\n✅ Vinho atualizado com sucesso!\n")
        else:
            print("\n⚠️ ID não encontrado.\n")
    except ValueError:
        print("\n❌ ID inválido.\n")

def deletar_vinho():
    try:
        id_vinho = int(input("\nDigite o ID do vinho que deseja deletar: "))
        global vinhos
        for v in vinhos:
            if v["id"] == id_vinho:
                vinhos.remove(v)
                salvar_vinhos(vinhos)
                print(f"\n🗑️ Vinho '{v['nome']}' removido com sucesso!\n")
                return
        print("\n⚠️ Nenhum vinho com esse ID foi encontrado.\n")
    except ValueError:
        print("\n❌ ID inválido.\n")

#(menu)
while True:
    print("""
==========================
🍷 ADEGA DE VINHOS 🍷
==========================
1 - Listar Vinhos
2 - Cadastrar Vinho
3 - Pesquisar Vinho
4 - Atualizar Vinho
5 - Deletar Vinho
0 - Sair
""")
    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        listar_vinhos()
    elif opcao == "2":
        cadastrar_vinho()
    elif opcao == "3":
        pesquisar_vinho()
    elif opcao == "4":
        atualizar_vinho()
    elif opcao == "5":
        deletar_vinho()
    elif opcao == "0":
        print("\nEncerrando o sistema... Até logo! 🍇")
        break
    else:
        print("\n⚠️ Opção inválida! Tente novamente.\n")
```
